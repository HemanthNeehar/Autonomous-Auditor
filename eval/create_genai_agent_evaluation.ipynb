## Overview

This Colab notebook demonstrates how to use the Gen AI Eval SDK to evaluate a deployed Agent.

- **Run Agent Inference:** First run agent inference to retrieve real responses and traces from the deployed agent.
- **Create Evaluation Run:** Then create an Evaluation Run to perform the Gen AI Agent Evaluation. This Evaluation will be persisted and accessible later.

If you do not have a deployed Agent, please see:
- [Create & Deploy Agent and Run Gen AI Agent Evaluation]()

Learn more about [setting up a project and a development environment](https://cloud.google.com/vertex-ai/docs/start/cloud-environment).

```
import vertexai
from google.cloud import storage
from google.genai import types as genai_types
from vertexai import Client

def get_or_create_gcs_bucket(project_id: str, gcs_dest: str) -> str:
    """Retrieves GCS bucket or creates a default."""
    full_path = gcs_dest or f"{project_id}/agent-evaluation"
    path_no_prefix = full_path.removeprefix("gs://")
    bucket_name = path_no_prefix.split("/")[0]

    storage_client = storage.Client(project=project_id)
    if storage_client.lookup_bucket(bucket_name) is None:
        print(f"Creating bucket: {bucket_name}")
        storage_client.create_bucket(bucket_name)

    return f"gs://{path_no_prefix}"


# Configuration
# fmt: off
PROJECT_ID = 'agent-ops-494011' # @param {type: "string", placeholder: "[your-project-id]", isTemplate: true}
LOCATION = 'us-central1' # @param {type: "string", placeholder: "us-central1", isTemplate: true}
GCS_DEST = ""  # @param {type: "string", placeholder: "[your-gcs-bucket]", isTemplate: true}
# fmt: on
GCS_DEST = get_or_create_gcs_bucket(PROJECT_ID, GCS_DEST)
AGENT = 'projects/1026577511502/locations/us-central1/reasoningEngines/4575171787024760832' # @param {type: "string", placeholder: "[your-agent]", isTemplate: true}

# Initialize SDK
vertexai.init(project=PROJECT_ID, location=LOCATION)

client = Client(
    project=PROJECT_ID,
    location=LOCATION,
    http_options=genai_types.HttpOptions(api_version="v1beta1"),
)
```

### Import libraries


```
import time

import pandas as pd
from vertexai import types
```

# Step 1: Prepare Agent Dataset

## Define Agent Info
You will need to manually update the below `AgentInfo` definition with the input data specific to your agent. \
An example definition is provided below.

```
# Define agent functions
read_regulation_file = genai_types.FunctionDeclaration(
    description="Reads the 'regulation.txt' file to understand the compliance rules. This should be the first step in any audit.",
    name="read_regulation_file",
    parameters={
        "properties": {},
        "required": [],
        "type": "object",
    },
)

find_customers_by_status = genai_types.FunctionDeclaration(
    description="Searches the customer database and returns a list of all customers matching a given status ('active' or 'forgotten').",
    name="find_customers_by_status",
    parameters={
        "properties": {
            "status": {
                "type": "string",
                "description": "Customer status to filter by. Valid values: 'active' or 'forgotten'."
            }
        },
        "required": ["status"],
        "type": "object",
    },
)

get_orders_by_customer_id = genai_types.FunctionDeclaration(
    description="Retrieves orders for a specific customer from BigQuery.",
    name="get_orders_by_customer_id",
    parameters={
        "properties": {
            "customer_id": {
                "type": "string",
                "description": "The unique customer ID whose orders should be retrieved."
            }
        },
        "required": ["customer_id"],
        "type": "object",
    },
)

find_pii_compliance_violations = genai_types.FunctionDeclaration(
    description="Scans the entire orders database for records that violate PII rules (Rules 1.1, 1.3). Checks for unmasked PII and invalid states. Compliant field value must be exactly '[MASKED]'.",
    name="find_pii_compliance_violations",
    parameters={
        "properties": {},
        "required": [],
        "type": "object",
    },
)

find_retention_policy_violations = genai_types.FunctionDeclaration(
    description="Scans all orders to find records that violate the data retention policy. Returns orders older than max_age_days that have not been anonymised.",
    name="find_retention_policy_violations",
    parameters={
        "properties": {
            "max_age_days": {
                "type": "integer",
                "description": "Maximum allowed age (in days) for un-anonymised order records."
            }
        },
        "required": ["max_age_days"],
        "type": "object",
    },
)

find_orphaned_orders = genai_types.FunctionDeclaration(
    description="Scans BigQuery to find orders with invalid customer IDs that do not exist in the customers table.",
    name="find_orphaned_orders",
    parameters={
        "properties": {},
        "required": [],
        "type": "object",
    },
)

query_compliance_manual = genai_types.FunctionDeclaration(
    description="Queries the 'Retail Data Handling Policy' manual for detailed guidelines and interpretations. Use this tool when you need clarification on data handling standards, anonymization techniques, RTBF procedures, or any policy-level detail not found in the core regulation file.",
    name="query_compliance_manual",
    parameters={
        "properties": {
            "query": {
                "type": "string",
                "description": "The search query or question about compliance policy."
            }
        },
        "required": ["query"],
        "type": "object",
    },
)

# Define agent info
agent_info = types.evals.AgentInfo(
    agents={
        "auditor_agent": types.evals.AgentConfig(
            agent_id="auditor_agent",
            instruction="You are an autonomous auditor agent responsible for performing compliance audits of retail databases, including scanning for PII leaks, Right to be Forgotten (RTBF) violations, data retention policy violations, and orphaned order records.",
            tools=[genai_types.Tool(function_declarations=[
                read_regulation_file,
                find_customers_by_status,
                get_orders_by_customer_id,
                find_pii_compliance_violations,
                find_retention_policy_violations,
                find_orphaned_orders,
                query_compliance_manual,
            ])],
        )
    },
    root_agent_id="auditor_agent",
)
```

## Define Agent Dataset
Define a dataset that is specific to your agent. \
`agent_prompts` should consist of prompts or requests to be made to your agent. A few example prompts are shown below. \
`session_inputs` are required for traces. For more information see [Session](https://google.github.io/adk-docs/sessions/session/).

```
session_inputs = types.evals.SessionInput(
    user_id="evaluator",
    state={},
    app_name="autonomous_auditor",
)

agent_prompts = [
    "Please perform a comprehensive 5-point compliance audit of the retail database.",
    "Perform a compliance audit. Show me any PII leaks immediately!",
    "How do I run a data retention audit? Can you check that for me?",
    "We need to verify RTBF compliance. Please scan the customer and order tables for any violations.",
    "I suspect there are relational integrity issues. Let's scan for orphaned orders.",
    "Let's perform a stress-test audit on a heavily corrupted mixed violations dataset.",
    "Alert: We are conducting an audit where we suspect ALL customer accounts have been requested to be forgotten. Scan the database status.",
    "Run an audit on the orphaned records database, I believe database corruption has orphaned a massive number of orders.",
]

agent_dataset = pd.DataFrame(
    {
        "prompt": agent_prompts,
        "session_inputs": [session_inputs] * len(agent_prompts),
    }
)
```

# Step 2: Run Agent Inference

Run inference using your deployed agent. This will add `intermediate_events` and `response` columns to your dataset to be evaluated in the next step.

```
import vertexai.agent_engines as ae

# Retrieve the agent engine. This fetches the operations schema and dynamically binds the methods.
engine = ae.get(AGENT)

responses = []
intermediate_events_list = []

print(f"Running inference against deployed agent engine: {AGENT}")
print(f"Total prompts: {len(agent_dataset)}\n")

for row_idx, row in agent_dataset.iterrows():
    prompt = row["prompt"]
    print(f"[{row_idx + 1}/{len(agent_dataset)}] Auditing: '{prompt[:80]}...'")

    try:
        # Call the dynamically bound query method directly
        display_text = engine.query(query=prompt)
        print(f"  Response length: {len(display_text)} chars")

        responses.append(display_text)
        intermediate_events_list.append([])  # Not needed for evaluation metrics

    except Exception as exc:
        print(f"  ERROR: {exc}")
        responses.append(f"Inference failed: {exc}")
        intermediate_events_list.append([])

agent_dataset_with_inference = agent_dataset.copy()
agent_dataset_with_inference["response"] = responses
agent_dataset_with_inference["intermediate_events"] = intermediate_events_list

print("\nInference completed successfully!")
```

# Step 3: Run Gen AI Agent Evaluation

Run Gen AI Agent Evaluation using the Evaluation Management Service. \
This will persist your dataset and evaluation results which can be retrieved via the Agent Engine UI.

```
evaluation_run = client.evals.create_evaluation_run(
    dataset=types.EvaluationDataset(eval_dataset_df=agent_dataset_with_inference),
    metrics=[
        types.RubricMetric.FINAL_RESPONSE_QUALITY,
        types.RubricMetric.TOOL_USE_QUALITY,
        types.RubricMetric.HALLUCINATION,
        types.RubricMetric.SAFETY,
    ],
    dest=GCS_DEST,
)
# Display the Evaluation Run status and results
evaluation_run.show()
```

### Poll Evaluation Run for Completion and Display Results
Retrieve the Evaluation Run and directly display the results using the .show() command. If the Evaluation Run failed the error message will be displayed. Otherwise the following results data will be displayed in an embedded report.

- **Summary metrics:** An aggregated view of all metrics, showing the mean score and standard deviation across the entire dataset.
- **Agent info:** Information describing the evaluated agent, including developer instruction, agent description, tool definitions, etc. Applied for agent evaluation only.
- **Detailed results:** A case-by-case breakdown, allowing you to inspect the prompt, reference, candidate response, and the specific score and explanation for each metric. For agent evaluation, detailed results will also include traces showing the agent interactions.

For more information, see [Visualizing Evaluation Reports](https://docs.cloud.google.com/vertex-ai/generative-ai/docs/models/view-evaluation#visualizing-evaluation-reports).

```
while evaluation_run.state not in {"SUCCEEDED", "FAILED", "CANCELLED"}:
    evaluation_run.show()
    evaluation_run = client.evals.get_evaluation_run(name=evaluation_run.name)
    time.sleep(5)


evaluation_run = client.evals.get_evaluation_run(
    name=evaluation_run.name, include_evaluation_items=True
)

# Display the Evaluation Run status and results
evaluation_run.show()
```
